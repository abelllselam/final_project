-> Cross Validation (k-Fold cross validation) - instead of train once and test once. We train and test multiple times. We use 10 folds split data into 10 parts, use 9 parts -> training and 1part testing and repeat this 10 times. Each time different part is used for testing. Cross validation evaluates the model multiple times on different splits of the data to get a more reliable performance estimate.

-> cross validation import

```python
from sklearn.model_selection import cross_val_score
```

-> Run cross validation on the Random Forest model. MSE shows how far predictions are from actual values (Penalizes large errors more because of squaring). So we use MSE. In Scikit-learn, all scores are designed so that higher = better. But in MSE lower is better. So it returns lower MSE so -100 is better than -200. Scikit-learn (A python library for machine learning that provides algorithms and tools) flips MSE to negative so it can follow its higher is better rule.

```python
cv_scores = cross_val_score(
    rf_model,
    X,
    y,
    cv=10,
    scoring="neg_mean_squared_error"
)
```

-> Output: The cross-validation results show some variation across folds, indicating that model performance changes depending on the data split. Some around: -300M to -900M
One very large: -1.47B

```python
cv_scores
```

-> convert these into RMSE (Root Mean Squared Error) so they are understandable. For the result below we have: Best case ≈ 18,400, Worst case ≈ 38,391, Most values ≈ 22k – 30k

```python
import numpy as np

cv_rmse = np.sqrt(-cv_scores)
cv_rmse
```

-> Let us look at the average RMSE

```python
cv_rmse.mean()
```

-> Hyperparameter Tuning (for Random Forest) - Adjusting model settings to improve performance (n_estimators → number of trees and max_depth → how deep trees go). Next, we are going to try many combinations and find the best one using GridSearchCV tool. GridSearchCV is a method that systematically tests multiple hyperparameter combinations using cross validation to find the best performing model.Total combinations that will be tested below (100 trees + depth 10, 100 trees + depth 20, 200 trees + depth 10, 200 trees + depth 20)

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20]
}
```

```python
grid_search = GridSearchCV(
    RandomForestRegressor(random_state=42),
    param_grid,
    cv=5,
    scoring="neg_mean_squared_error"
)
```

```python
grid_search.fit(X_train, y_train)
```

-> Explaining the tuning (GridSearch) - Best model is RandomForestRegressor(max_depth=20, n_estimators=200, random_state=42), meaning that the best model used 200 trees and each tree can grow up to 20 levels deep. Running below just to show the best setting:

```python
grid_search.best_params_
```

-> Explaing the above result: I used GridSearchCV (Grid Search Cross Validation) to tune the hyperparameters of my Random Forest model. The best configuration was 200 trees with a maximum depth of 20, which provided the best performance based on cross-validation.

-> Now let us test the optimized model:

```python
best_model = grid_search.best_estimator_

best_pred = best_model.predict(X_test)

best_mse = mean_squared_error(y_test, best_pred)

best_rmse = np.sqrt(best_mse)

print(best_rmse)
```

***Next we will start with creating catagories***
Splits prices into 3 groups
Converts regression → classification (Low / Medium / High price). Instead of using a fixed price data we are using boundaries based on data distribution. I used quantile-based binning to divide house prices into three equal groups. The thresholds are automatically determined based on the distribution of the data.

```python
df['PriceCategory'] = pd.qcut(df['SalePrice'], q=3, labels=['Low', 'Medium', 'High'])
```

```python
pd.qcut(df['SalePrice'], q=3)
```

-> What the result shows:
| Category   | Price Range       |
| ---------- | ----------------- |
| **Low**    | 12,789 → 139,000  |
| **Medium** | 139,000 → 190,000 |
| **High**   | 190,000 → 755,000 |

I used quantile-based binning to divide the data into three equal groups. The thresholds were approximately $139,000 and $190,000, separating low, medium, and high-priced houses. The dataset was divided into three equal price groups based on distribution.

-> Next, let us prepare the data for classification:

```python
y_class = df["PriceCategory"] #Low / Medium / High target

X_class = df.drop(["SalePrice", "PriceCategory"], axis=1) #all house features

X_class = pd.get_dummies(X_class) #only converts the features, not the target
```

```python
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class,
    y_class,
    test_size=0.2,
    random_state=42
)
# _c means classification
# 80% training
# 20% testing
```

-> Import the classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

-> create a Random Forest model for classification

```python
rf_classifier = RandomForestClassifier(random_state=42)
```

-> Train the classifier: fit() trains the classifier using:

X_train_c = house features
y_train_c = correct labels: Low, Medium, High

```python
rf_classifier.fit(X_train_c, y_train_c)
```

-> Now make category prediction

```python
y_pred_c = rf_classifier.predict(X_test_c)
```

```python
y_pred_c[:10]
```

-> show classification metrics, It will show Accuracy, precision, recall and F1-score for Low, Medium and High. Not just predictions, but how good they are

```python
from sklearn.metrics import classification_report

print(classification_report(y_test_c, y_pred_c))
```

-> Explaining the result:
| Metric    | What it measures                          |
| --------- | ----------------------------------------- |
| Accuracy  | Overall correctness                       |
| Precision | How correct your predictions are |
| Recall    | How many actual positives you found       |
| F1-score  | Balance of precision and recall           |

- For the High, the model saysit is correct 95% of the time (precision), Out of all the actual values the model found 0.91 or 91% in Low classification, and On the F1-score (Harmonic Mean of Precision and Recall) which it is 0.88 which is a good balance between finding and being correct. The support shows, Number of actual samples in each class, There are 223 high priced houses in test data.

- I evaluated my classification model using accuracy, precision, recall, and F1-score. Accuracy measures overall correctness, while precision and recall evaluate performance for each class. F1-score provides a balanced measure, and support indicates the number of samples per class.

- Precision measures how accurate the model’s predictions are for each class. For example, when the model predicts a house as High, it is correct 95% of the time.
- Recall? Out of all actual values, how many did the model correctly find?
- F1-score (Harmonic Mean of Precision and Recall), A balance between precision and recall, Sometimes: Precision is high and Recall is low; F1 combines both into one number

- Putting it all together: 
    I evaluated my classification model using accuracy, precision, recall, and F1-score. Accuracy measures overall correctness. Precision measures how accurate the model’s predictions are for each class. Recall measures how well the model identifies all actual instances of each class. F1-score provides a balanced measure by combining precision and recall. My model achieved an accuracy of 88%, with high precision and recall across all classes. This indicates that the model performs well in correctly classifying low, medium, and high-priced houses.

-> To improve the classification, let us try the Hyperparameter tuning for classifier

```python
param_grid_clf = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20]
}
```

-> Create GridSearch for the classifier

```python
grid_clf = GridSearchCV(
    RandomForestClassifier(random_state=42), #classification model
    param_grid_clf, #settings to try
    cv=5, #5-fold cross validation
    scoring="accuracy" #optimize for accuracy
)
```

-> Run hyperparameter tuning for the classifier

```python
grid_clf.fit(X_train_c, y_train_c)
```

```python
grid_clf.best_params_
```

-> The next test tuned classifier will show whether tuning improved accuracy

```python
best_clf = grid_clf.best_estimator_

y_pred_best_c = best_clf.predict(X_test_c)

print(classification_report(y_test_c, y_pred_best_c))
```
